```python
X = df.drop("Class", axis=1)
y = df["Class"]
```

```python
from sklearn.model_selection import train_test_split

# Identify rows where y is NaN
nan_mask = y.isna()

# Drop rows with NaN from X and y
X_cleaned = X[~nan_mask]
y_cleaned = y[~nan_mask]

X_train, X_test, y_train, y_test = train_test_split(
    X_cleaned, y_cleaned, test_size=0.2, stratify=y_cleaned, random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
import torch
import torch.nn as nn

class CreditNet(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(32, 1)
        )

    def forward(self, x):
        return self.net(x)
```

```python
import torch
import pandas as pd
from sklearn.metrics import roc_auc_score
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

X_train_t = torch.tensor(X_train, dtype=torch.float32)
y_train_t = torch.tensor(y_train.values, dtype=torch.float32).unsqueeze(1)
X_test_t = torch.tensor(X_test, dtype=torch.float32)

model = CreditNet(X_train.shape[1])
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

# Calculate class weights and define pos_weight
classes = np.array([0,1])
weights = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
pos_weight = torch.tensor(weights[1], dtype=torch.float32)

criterion = torch.nn.BCEWithLogitsLoss(pos_weight=pos_weight)


for epoch in range(15):
    model.train()
    optimizer.zero_grad()
    logits = model(X_train_t)
    loss = criterion(logits, y_train_t)
    loss.backward()
    optimizer.step()

model.eval()
with torch.no_grad():
    probs = torch.sigmoid(model(X_test_t)).numpy().ravel()

print("Baseline AUC:", roc_auc_score(y_test, probs))
```

```python
import numpy as np

preds = (probs > 0.5).astype(int)
confidence = np.maximum(probs, 1 - probs)
```

```python
confident_wrong = (confidence > 0.9) & (preds != y_test.values)

print("High-confidence wrong predictions:", confident_wrong.sum())
print("Total test samples:", len(y_test))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Binary predictions (threshold = 0.5)
y_pred_label = (mc_mean >= 0.5).astype(int)

# Correct / incorrect masks
correct_mask = (y_pred_label == y_test.values)
incorrect_mask = ~correct_mask

plt.figure()
plt.hist(
    mc_std[correct_mask],
    bins=50,
    alpha=0.6,
    label="Correct predictions"
)
plt.hist(
    mc_std[incorrect_mask],
    bins=50,
    alpha=0.6,
    label="Incorrect predictions"
)

plt.xlabel("Epistemic Uncertainty (MC Dropout Std)")
plt.ylabel("Number of Samples")
plt.legend()
plt.title("Uncertainty Distribution of Predictions")
plt.show()
```

```python
# Sort samples by uncertainty (low → high)
sorted_indices = np.argsort(mc_std)

sorted_true = y_test.values[sorted_indices]
sorted_pred = y_pred_label[sorted_indices]

risks = []
coverages = []

n = len(sorted_true)

for k in range(1, n + 1):
    # Keep k lowest-uncertainty samples
    kept_true = sorted_true[:k]
    kept_pred = sorted_pred[:k]

    risk = np.mean(kept_pred != kept_true)
    coverage = k / n

    risks.append(risk)
    coverages.append(coverage)

plt.figure()
plt.plot(coverages, risks)
plt.xlabel("Coverage (Fraction of Predictions Kept)")
plt.ylabel("Risk (Error Rate)")
plt.title("Risk–Coverage Curve (MC Dropout)")
plt.show()
```

```python
import numpy as np
import torch

model.train()  # IMPORTANT: activate dropout at inference

T = 50  # number of MC samples
mc_preds = []

with torch.no_grad():
    for _ in range(T):
        p = torch.sigmoid(model(X_test_t)).cpu().numpy().ravel()
        mc_preds.append(p)

mc_preds = np.array(mc_preds)

mc_mean = mc_preds.mean(axis=0)   # predictive mean
mc_std  = mc_preds.std(axis=0)    # epistemic uncertainty
```

```python
# Binary predictions
preds_mc = (mc_mean > 0.5).astype(int)

# Model confidence (distance from 0.5)
confidence = np.maximum(mc_mean, 1 - mc_mean)
```

```python
y_true = y_test.values if hasattr(y_test, "values") else y_test

confident_wrong = (
    ((mc_mean > 0.9) & (y_true == 0)) |
    ((mc_mean < 0.1) & (y_true == 1))
)

print("High-confidence wrong predictions:", confident_wrong.sum())
print("Total test samples:", len(y_true))
```

```python
def expected_calibration_error(y_true, probs, n_bins=10):
    bins = np.linspace(0, 1, n_bins + 1)
    ece = 0.0

    for i in range(n_bins):
        mask = (probs > bins[i]) & (probs <= bins[i+1])
        if mask.sum() == 0:
            continue

        acc = (y_true[mask] == (probs[mask] > 0.5)).mean()
        conf = probs[mask].mean()
        ece += (mask.sum() / len(probs)) * abs(acc - conf)

    return ece
```

```python
ece = expected_calibration_error(y_true, mc_mean)
print("Expected Calibration Error (ECE):", round(ece, 4))
```

```python
model.eval()  # temperature scaling uses deterministic outputs

with torch.no_grad():
    logits = model(X_test_t).cpu().numpy().ravel()

y_true = y_test.values if hasattr(y_test, "values") else y_test

```

```python
import torch.nn as nn

class TemperatureScaler(nn.Module):
    def __init__(self):
        super().__init__()
        self.temperature = nn.Parameter(torch.ones(1))

    def forward(self, logits):
        return logits / self.temperature
```

```python
import torch
import numpy as np

scaler = TemperatureScaler()
optimizer = torch.optim.LBFGS([scaler.temperature], lr=0.01, max_iter=50)
criterion = nn.BCEWithLogitsLoss()

logits_t = torch.tensor(logits, dtype=torch.float32)
labels_t = torch.tensor(y_true, dtype=torch.float32)

def closure():
    optimizer.zero_grad()
    loss = criterion(scaler(logits_t), labels_t)
    loss.backward()
    return loss

optimizer.step(closure)

print("Optimal temperature:", scaler.temperature.item())
```

```python
with torch.no_grad():
    scaled_logits = scaler(logits_t)
    probs_temp = torch.sigmoid(scaled_logits).numpy()
```

```python
ece_before = expected_calibration_error(y_true, mc_mean)
ece_after  = expected_calibration_error(y_true, probs_temp)

print("ECE before temperature scaling:", round(ece_before, 4))
print("ECE after temperature scaling:", round(ece_after, 4))
```

```python
from sklearn.metrics import accuracy_score, roc_auc_score

preds_before = (mc_mean > 0.5).astype(int)
preds_after  = (probs_temp > 0.5).astype(int)

print("Accuracy before:", accuracy_score(y_true, preds_before))
print("Accuracy after:", accuracy_score(y_true, preds_after))

print("AUC before:", roc_auc_score(y_true, mc_mean))
print("AUC after:", roc_auc_score(y_true, probs_temp))
```

```python
import numpy as np

def reliability_curve(y_true, probs, n_bins=10):
    bins = np.linspace(0, 1, n_bins + 1)
    bin_centers = []
    bin_acc = []
    bin_conf = []
    bin_counts = []

    for i in range(n_bins):
        mask = (probs > bins[i]) & (probs <= bins[i+1])
        if mask.sum() == 0:
            continue

        acc = (y_true[mask] == (probs[mask] > 0.5)).mean()
        conf = probs[mask].mean()

        bin_centers.append((bins[i] + bins[i+1]) / 2)
        bin_acc.append(acc)
        bin_conf.append(conf)
        bin_counts.append(mask.sum())

    return np.array(bin_centers), np.array(bin_acc), np.array(bin_conf), np.array(bin_counts)
```

```python
from sklearn.metrics import accuracy_score
import numpy as np

def risk_coverage_curve(y_true, probs, uncertainty, thresholds):
    risks = []
    coverages = []

    preds = (probs > 0.5).astype(int)

    for t in thresholds:
        keep = uncertainty <= t

        if keep.sum() == 0:
            continue

        acc = accuracy_score(y_true[keep], preds[keep])
        risk = 1 - acc
        coverage = keep.mean()

        risks.append(risk)
        coverages.append(coverage)

    return np.array(coverages), np.array(risks)
```

```python
from sklearn.metrics import accuracy_score
import numpy as np

def risk_coverage_curve(y_true, probs, uncertainty, thresholds):
    risks = []
    coverages = []

    preds = (probs > 0.5).astype(int)

    for t in thresholds:
        keep = uncertainty <= t

        if keep.sum() == 0:
            continue

        acc = accuracy_score(y_true[keep], preds[keep])
        risk = 1 - acc
        coverage = keep.mean()

        risks.append(risk)
        coverages.append(coverage)

    return np.array(coverages), np.array(risks)
```

```python
baseline_preds = (mc_mean > 0.5).astype(int)
baseline_acc = accuracy_score(y_true, baseline_preds)
baseline_risk = 1 - baseline_acc

print("Baseline coverage:", 1.0)
print("Baseline risk:", round(baseline_risk, 4))
```

```python
from sklearn.metrics import recall_score

def recall_coverage_curve(y_true, probs, uncertainty, thresholds):
    recalls = []
    coverages = []

    preds = (probs > 0.5).astype(int)

    for t in thresholds:
        keep = uncertainty <= t
        if keep.sum() == 0:
            continue

        rec = recall_score(y_true[keep], preds[keep])
        coverages.append(keep.mean())
        recalls.append(rec)

    return np.array(coverages), np.array(recalls)
```

```python
thresholds = np.linspace(mc_std.min(), mc_std.max(), 100)

coverage_r, recall = recall_coverage_curve(
    y_true, mc_mean, mc_std, thresholds
)

plt.figure(figsize=(6, 5))
plt.plot(coverage_r, recall, marker="o")
plt.xlabel("Coverage")
plt.ylabel("Recall")
plt.title("Recall vs Coverage")
plt.grid(True)
plt.show()
```
